# Unit I: Climate Change Module

## Reproducing the hockey stick, and checking whether we got it right

Our target is the result reported by [Mann et al (1998)](https://doi.org/10.1038/33859):
that recent warming stands outside the range of natural variability over the past several
centuries. You will rebuild that picture from current observations and ice-core records,
in the style of NASA's [vital signs](https://climate.nasa.gov/vital-signs).

### How this course works, and why

You have a language model that will write parsing code for you. It is good at it. So a
fair question on day one is what is left for you to do.

Consider what happened every previous time computing got dramatically easier. Punchcards
gave way to interactive screens. Assembly gave way to compilers. C gave way to Python.
Each of those was an enormous efficiency gain, and not one of them resulted in fewer
people programming, or in programmers spending less time building software. Every single
time, the gain was spent on attempting more, not on going home early.

The same pattern shows up in the hardware. Processors today are orders of magnitude more
energy-efficient per computation than those of thirty years ago. Total electricity spent
on computing has gone up, not down. Efficiency did not reduce consumption; it expanded
what people tried to compute.

So if a language model makes you ten times faster, history offers no support at all for
the idea that you will spend one tenth of the time. You will do more. You will attempt
analyses that were previously out of reach — which is exactly what this course asks of
you.

Two things follow, and they pull in opposite directions:

**We are not going to teach you to write every command from memory.** Memorizing library
syntax was never the skill; it was the toll you paid to get to the actual work. That toll
is now much lower, so we can spend the time on harder and more interesting questions.

**This does not mean you can paste an assignment into a model, type "go", and collect an
A.** Working with a model takes real time and real judgment, in the same way that writing
Python takes time — just less of it than writing C. The effort moves; it does not vanish.
What it moves toward is the question this module is built around, the one scientists have
always had to answer about code they did not write themselves:

> **How do I know these numbers are right?**

That question long predates language models. Parsers have always misread missing values,
silently coerced types, loaded one file out of twelve, and mislabeled columns. Humans have
always done this. Models do it faster, more fluently, and in more places at once.

### Task 3.2: Measure what it cost

Run the code it wrote using the harness below, which reports peak memory and wall time.

Your JupyterHub session has roughly **4 GB of RAM**. Note what happens.

````python
import resource, time, contextlib

@contextlib.contextmanager
def measure(label):
    t0 = time.time()
    try:
        yield
    finally:
        peak = resource.getrusage(resource.RUSAGE_SELF).ru_maxrss / 1024**2
        print(f"{label}: peak memory {peak:.2f} GB, elapsed {time.time() - t0:.1f}s")

# Example:
# with measure("pandas"):
#     result = ...
````
